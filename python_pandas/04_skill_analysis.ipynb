## Analysis 1: Period-Level Skill Summary

This analysis summarizes average skill scores by period.  
It helps compare how skill requirements change across periods.

```python
skill_period_summary = df_skill.groupby("period").agg(
    average_personal_skill=("personal_skill", "mean"),
    average_communication_skill=("communication_skill", "mean"),
    average_problem_solving_skill=("problem_solving_skill", "mean"),
    average_social_skill=("social_skill", "mean"),
    average_technical_skill=("technical_skill", "mean")
).reset_index()

skill_period_summary
```

## Analysis 2: Overall Average Skill Scores

This analysis calculates the overall average score for each skill type across all periods and occupation groups.

```python
overall_skill_scores = df_skill[
    ["personal_skill", "communication_skill", "problem_solving_skill", "social_skill", "technical_skill"]
].mean().sort_values(ascending=False)

overall_skill_scores
```

## Analysis 3: Top Occupation Groups by Technical Skill

This analysis identifies the occupation groups with the highest technical skill scores.

```python
top_technical_skill = df_skill[
    ["period", "occupation_group", "technical_skill"]
].sort_values(by="technical_skill", ascending=False)

top_technical_skill.head(10)
```

## Analysis 4: Top Occupation Groups by Problem-Solving Skill

This analysis identifies the occupation groups with the highest problem-solving skill scores.

```python
top_problem_solving_skill = df_skill[
    ["period", "occupation_group", "problem_solving_skill"]
].sort_values(by="problem_solving_skill", ascending=False)

top_problem_solving_skill.head(10)
```

## Analysis 5: Dominant Skill by Occupation Group

This analysis identifies the highest-rated skill category for each occupation group and period.  
It helps understand which skill type is most prominent across different occupation groups.

```python
skill_columns = [
    "personal_skill",
    "communication_skill",
    "problem_solving_skill",
    "social_skill",
    "technical_skill"
]

df_skill["dominant_skill"] = df_skill[skill_columns].idxmax(axis=1)

df_skill["dominant_skill"] = (
    df_skill["dominant_skill"]
    .str.replace("_skill", "", regex=False)
    .str.replace("_", " ")
    .str.title()
)

dominant_skill_by_occupation = df_skill[
    ["period", "occupation_group", "dominant_skill"]
]

dominant_skill_by_occupation.head(10)
```

```python
dominant_skill_distribution = (
    df_skill["dominant_skill"]
    .value_counts()
    .reset_index()
)

dominant_skill_distribution.columns = ["dominant_skill", "count"]

dominant_skill_distribution
```

## Key Insights

- Social skill has the highest overall average score across the dataset.
- Technical skill and personal skill also appear as relatively strong skill categories.
- Communication skill has the lowest overall average score in this dataset.
- Several occupation groups show high technical skill scores, especially in agriculture, service/sales, technician-related and craft-related groups.
- Problem-solving skill appears strongly in office, agriculture, service/sales, technician-related and machine operator groups.
- Since the skill scores appear to be distributed as ranked values, total skill score is not useful for comparison. Instead, dominant skill analysis provides a clearer view of the strongest skill category by occupation group.
